**4.** Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
A = np.random.randint(1, 10, size=(8, 3))
print(A)
B = np.random.randint(1,10, size=(2, 2))
print(B)
result = []

for x in A:
    if all(any(element in x for element in y) for y in B):
        result.append(x)
result = np.array(result)
print(f"result:\n {result}")
            
        
```

**5.** Дана 10x3 матрица, найти строки из неравных значений (например строка [2,2,3] остается, строка [3,3,3] удаляется)

```python
array = np.random.randint(1, 3, (10,3))
print(array)

result = []
for row in array:
    if not (row == row[0]).all():
        result.append(row)
 
result = np.array(result)
print(f"result: \n{result}")
```

**6.** Дан двумерный массив. Удалить те строки, которые повторяются

* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
print(X)

result=1
for element in X.diagonal():
    if (element != 0):
        result*=element

print(result)

# no NumPy
result = 1

for i in range(min(len(X), len(X[0]))):
    if X[i][i] != 0:
        result *= X[i][i]

print(result)
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
#No NumPy
x = [4, 4, 1, 1, 1, 7, 7, 4]
counts=[]
ammount=[]
for element in x:
    if counts:
        if counts[-1]==element:
            ammount[-1]+=1
        else:
            counts.append(element)
            ammount.append(1)
    else: 
        counts.append(element)
        ammount.append(1)


result = (counts, ammount)
print("My result:")
print(result)
# numpy gpt
x = np.array([4, 4, 1, 1, 1, 7, 7, 4])

changes = np.where(x[1:] != x[:-1])[0]+1

starts = np.r_[0, changes]
ends = np.r_[changes, len(x)]

length_of = ends - starts
values = x[starts]

resultgpt = (values, length_of)
print(f"resultgpt: \n {resultgpt}")

```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import math
from scipy.spatial.distance import cdist
# No NumPy
X = [
    [0, 0],
    [3, 4]
]

Y = [
    [0, 4],
    [3, 0],
    [6, 8]
]

def my_distance(X, Y):
    result = [[0] * len(Y) for _ in range(len(X))]

    a = 0
    b = 0

    for x in X:
        for y in Y:
            distance = 0

            for i in range(len(x)):
                distance += (y[i] - x[i]) ** 2

            result[a][b] = math.sqrt(distance)
            b += 1

        b = 0
        a += 1

    return result


print(my_distance(X, Y))



# NumPy

X = np.array([
    [0, 0],
    [3, 4]
])

Y = np.array([
    [0, 4],
    [3, 0],
    [6, 8]
])


def numpy_distance(X, Y):
    return np.sqrt(np.sum((X[:, np.newaxis, :] - Y) ** 2, axis=2))


print(numpy_distance(X, Y))
print(cdist(X, Y))
print(np.allclose(numpy_distance(X, Y), cdist(X, Y)))



# Speed test

X_big = np.random.random((1000, 10))
Y_big = np.random.random((1000, 10))

X_big_list = X_big.tolist()
Y_big_list = Y_big.tolist()

%timeit my_distance(X_big_list, Y_big_list)
%timeit numpy_distance(X_big, Y_big)
%timeit cdist(X_big, Y_big)

# Python: 3.34 s
# NumPy: 136 ms
# scipy.spatial.distance.cdist: 11.4 ms
#
# NumPy примерно в 24.6 раза быстрее Python.
# cdist примерно в 11.9 раза быстрее NumPy.
# cdist примерно в 293 раза быстрее Python.
```
